```python
def heapify(arr, n, i):
    """
    Ensures the sub-tree rooted at index 'i' satisfies the Max-Heap property.
    n: size of the heap
    i: current index to sift down
    """
    largest = i          # Initialize largest as root
    left = 2 * i + 1     # Left child index
    right = 2 * i + 2    # Right child index

    # If left child exists and is greater than root
    if left < n and arr[left] > arr[largest]:
        largest = left

    # If right child exists and is greater than current largest
    if right < n and arr[right] > arr[largest]:
        largest = right

    # If the largest is not the root, swap and continue heapifying down
    if largest != i:
        arr[i], arr[largest] = arr[largest], arr[i]  # Swap
        heapify(arr, n, largest)                      # Recursively heapify the affected sub-tree

def heapSort(arr):
    n = len(arr)

    # Step 1: Build a Max-Heap
    # Start from the last non-leaf node up to the root (index 0)
    for i in range(n // 2 - 1, -1, -1):
        heapify(arr, n, i)

    # Step 2: Extract elements from the heap one by one
    for i in range(n - 1, 0, -1):
        # Move current root (maximum element) to the end
        arr[0], arr[i] = arr[i], arr[0]
        
        # Call heapify on the reduced heap (size 'i')
        heapify(arr, i, 0)

# --- Test Case ---
arr = [4, 10, 3, 5, 1]
print(f"Original Array: {arr}")

heapSort(arr)
print(f"Sorted Array:   {arr}")
# Output: Sorted Array: [1, 3, 4, 5, 10]
```
