```python
num = [1,2,3,3,4,5,6,6,7,8,9,10]
for i in range(len(num)):
    if num[i] % 2 == 0:
        print(num[i],"is even")
    else:
        print(num[i],"is odd")
```

```python
num=[1,2,3,3,4,5,6,6,7,8,9,10]
while num:
    n=num.pop()
    if n%2==0:
        print(n,"is even")
    else:
        print(n,"is odd")
```
