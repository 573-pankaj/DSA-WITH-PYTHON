# finding maximum and minimum value

```python
data = [1,2,3,4,4,5,56,6,65,34,23,566,53,23,43,2,454,66,4,2345,5,5,6,7]
```

```python
max(data), min(data)
```

```python
np.mean(data)
```

```python
def rangef(data):
    maxV = max(data)
    minV = min(data)
    return(maxV-minV)
```

```python
rangeV = rangef(data)
```

```python
print(rangeV)
```
