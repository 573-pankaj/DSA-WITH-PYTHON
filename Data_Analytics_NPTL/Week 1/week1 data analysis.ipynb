```python
evens = [x for x in range(20) if x%2==0]   # trick 3
```

```python
print(evens)
```

```python
# trick 4
X = []
for x in range(1,21):
    X.append(x)
```

```python
print(X)
```

```python
def square(n):
    return n**2
def even(n):
    return n%2==0
def multiply(x,y):
    return x*y
```

```python
square = list(map(square,X))    # map()
print(square)
```

```python
evens = list(filter(even,X))  # filter()
print(evens)
```

```python
from functools import reduce
```

```python
mul = reduce(multiply,X)
print(mul)
```

```python
squares = list(map(lambda x:x**2,X))
print(squares)
```
