### Linked List Append Dry Run

**Initial State:** `head = None`

| Action | `data` | List Empty Check (`head is None`) | Traversal Log | Pointer Updates | Resulting Chain |
| :--- | :---: | :--- | :--- | :--- | :--- |
| **Step 1** | `10` | True | N/A | `head = Node(10)` | `[10] -> None` |
| **Step 2** | `20` | False | Start at `10`. `10.next` is `None`. Stop. | `Node(10).next = Node(20)` | `[10] -> [20] -> None` |
| **Step 3** | `30` | False | Start at `10`. Move to `20`. `20.next` is `None`. Stop. | `Node(20).next = Node(30)` | `[10] -> [20] -> [30] -> None` |

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
        
class LinkedListVisualizer:
    def __init__(self):
        self.head = None
    def append(self,data):
        new_node = Node(data)
        if self.head is None:
            self.head = new_node
            print(f"Action : List is empty . Node({data}) become the new node.")
            self.display()
            return

        current = self.head
        print(f"Action : List is not Empty.Starting traversal at head Node({current.data}).")
        while current.next is not None:
            print(f"Traversing: Node({current.data}).next is not None. Moving to Node({current.next.data}).")
            current = current.next

        current.next = new_node
        print(f"Action: Reached the end. Linking Node({current.data}).next to new Node({data}).")
        self.display()

    def display(self):
        current = self.head
        elements = []
        while current is not None:
            elements.append(f"[{current.data}]")
            current = current.next
        print(" Current Linked List : " + "--> ".join(elements) + "--> None")

    def traversal(self):
        if not self.head:
            print("SLL is Empty")
        else:
            current = self.head
            while current is not None:
                print(current.data,end = " " )
                current = current.next
            print()

        
        
        
        
```

```python
ll = LinkedListVisualizer()
ll.append(10)
ll.append(20)
ll.append(30)
ll.append(37)
ll.append(305)
ll.append(307)
ll.traversal()
```
