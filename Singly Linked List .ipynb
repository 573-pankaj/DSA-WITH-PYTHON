### Ground-Zero Guide: Linked List Insertions (Beginning, Middle, End)

#### 1. The Building Blocks (`Node` and `LinkedList.__init__`)
* `self.data = data`: Stores the actual payload inside the node object.
* `self.next = None`: Initializes the reference pointer to `None` because a freshly created node is not connected to anything yet.
* `self.head = None`: Inside `LinkedList`, `head` tracks the entry point of the chain. When `head` is `None`, the list is empty.

---

#### 2. Insert at the Beginning (`insert_at_beginning`) — $O(1)$ Time
Inserting at the front requires no traversal because we already have direct access to `self.head`.

1. `new_node = Node(data)`: Allocates a new node in memory.
2. `new_node.next = self.head`: **Crucial Order!** Before changing `self.head`, we make the new node point to whatever `self.head` is currently pointing to. If you update `self.head` first, you lose the reference to the rest of your list forever.
3. `self.head = new_node`: Moves the `head` label to the new front node.

---

#### 3. Insert at the End (`insert_at_end`) — $O(N)$ Time
Because a singly linked list only knows where the `head` is, we must walk the chain to find the last node.

1. `if self.head is None:`: Edge case check. If the list is empty, there is no last node to traverse to, so `self.head = new_node` sets our new node as the first and only node.
2. `current = self.head`: Creates a temporary walker pointer starting at the first node. Never move `self.head` itself during traversal, or you will lose the start of your list.
3. `while current.next is not None:`: Loops until `current` stands on the node whose `.next` is `None` (the tail node).
4. `current.next = new_node`: Attaches the new node to the end of the chain.

---

#### 4. Insert in the Middle (`insert_at_position`) — $O(N)$ Time
To insert a node at index `P`, we must stop at index `P - 1` (the predecessor node) so we can splice the new node in between `P - 1` and `P`.

1. `if position == 0:`: Handled immediately by calling `self.insert_at_beginning(data)`.
2. `while current is not None and current_index < position - 1:`: Moves `current` forward until it reaches the node **immediately before** the target index.
3. `if current is None:`: If the user asked for index `10` in a 3-element list, `current` walks off the end of the list and becomes `None`. This prevents a `NoneType` crash.
4. `new_node.next = current.next`: **Step A of splicing.** The new node grabs onto the node that currently sits at `position` (`current.next`).
5. `current.next = new_node`: **Step B of splicing.** The predecessor node (`current`) lets go of the old node and points to `new_node`.

---

#### 5. Step-by-Step Memory Dry Run

| Operation Call | Initial List State | Key Pointer Actions | Resulting List State |
| :--- | :--- | :--- | :--- |
| **`insert_at_end(20)`** | `None` | `head` is `None` $\rightarrow$ `head = Node(20)` | `[20] -> None` |
| **`insert_at_end(40)`** | `[20] -> None` | `current` stops at `20`. Sets `Node(20).next = Node(40)` | `[20] -> [40] -> None` |
| **`insert_at_beginning(10)`** | `[20] -> [40] -> None` | `Node(10).next = Node(20)`, then `head = Node(10)` | `[10] -> [20] -> [40] -> None` |
| **`insert_at_position(30, 2)`** | `[10] -> [20] -> [40] -> None` | Target is index `2`. Stop at index `1` (`Node(20)`).<br>1. `Node(30).next = Node(40)`<br>2. `Node(20).next = Node(30)` | `[10] -> [20] -> [30] -> [40] -> None` |

---

#### 6. Complexity Summary

| Operation | Time Complexity | Auxiliary Space | Why? |
| :--- | :---: | :---: | :--- |
| **Insert at Beginning** | $O(1)$ | $O(1)$ | Direct pointer reassignment at `self.head`; zero traversal. |
| **Insert in Middle (Pos $K$)** | $O(K)$ / $O(N)$ | $O(1)$ | Must traverse $K-1$ nodes from `head` to reach the predecessor. |
| **Insert at End** | $O(N)$ | $O(1)$ | Must traverse all $N$ nodes from `head` to locate the tail node. |

```python
class Node:
    def __init__(self, data):
        self.data = data      # Stores the value of the node
        self.next = None      # Pointer to the next node (initially None)

class LinkedList:
    def __init__(self):
        self.head = None      # Points to the first node in the linked list

    # 1. INSERT AT THE BEGINNING (First Position)
    def insert_at_beginning(self, data):
        new_node = Node(data)         # Step 1: Create the new node
        new_node.next = self.head     # Step 2: Point new node's next to current head
        self.head = new_node          # Step 3: Update head to be the new node

    # 2. INSERT AT THE END (Last Position)
    def insert_at_end(self, data):
        new_node = Node(data)         # Step 1: Create the new node
        
        if self.head is None:         # Step 2: If list is empty, make new node the head
            self.head = new_node
            return
            
        current = self.head           # Step 3: Start a temporary pointer at the head
        while current.next is not None: # Step 4: Traverse until the last node
            current = current.next
            
        current.next = new_node       # Step 5: Link the last node to the new node

    # 3. INSERT AT A SPECIFIC POSITION (Middle / 0-Based Index)
    def insert_at_position(self, data, position):
        if position < 0:              # Guard clause for invalid negative index
            print("Invalid position!")
            return
            
        if position == 0:             # Step 1: If index is 0, reuse insert_at_beginning
            self.insert_at_beginning(data)
            return
            
        new_node = Node(data)         # Step 2: Create the new node
        current = self.head           # Step 3: Start traversal from the head
        current_index = 0
        
        # Step 4: Stop at the node JUST BEFORE the target position (position - 1)
        while current is not None and current_index < position - 1:
            current = current.next
            current_index += 1
            
        # Step 5: Check if position was out of bounds
        if current is None:
            print(f"Position {position} is out of bounds!")
            return
            
        # Step 6: Rewire pointers (Order matters!)
        new_node.next = current.next  # Connect new node to the rest of the chain first
        current.next = new_node       # Connect previous node to the new node

    # HELPER: Display the Linked List
    def display(self):
        current = self.head
        elements = []
        while current is not None:
            elements.append(f"[{current.data}]")
            current = current.next
        print(" -> ".join(elements) + " -> None")


# --- Execution & Testing ---
ll = LinkedList()

print("1. Inserting 20 and 40 at the end:")
ll.insert_at_end(20)
ll.insert_at_end(40)
ll.display()

print("\n2. Inserting 10 at the beginning (Index 0):")
ll.insert_at_beginning(10)
ll.display()

print("\n3. Inserting 30 in the middle (Index 2):")
ll.insert_at_position(30, 2)
ll.display()
```
