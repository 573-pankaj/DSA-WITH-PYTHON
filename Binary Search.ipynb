# Lec 46 : Introduction to Binary Search

```python
arr = [1, 2, 3, 4, 5, 6, 7, 8, 9]
target = 68

def BinarySearch(arr, target):
    n = len(arr)
    low = 0
    high = n - 1
    
    # Use a while loop so low and high can shrink inward
    while low <= high:
        mid = (low + high) // 2
        
        # Case 1: Target found at mid index
        if arr[mid] == target:
            return f"Target element found at index {mid}"
            
        # Case 2: If target is SMALLER than mid element, it's in the left half
        # So we move 'high' down to mid - 1
        elif arr[mid] > target:
            high = mid - 1
            
        # Case 3: If target is LARGER than mid element, it's in the right half
        # So we move 'low' up to mid + 1
        else:
            low = mid + 1
            
    return f"Element not found {-1}"

print(BinarySearch(arr, target))
# Output: Target element found at index 5
```

```python
# Recursive Solutions
arr = [1, 2, 3, 4, 5, 6, 7, 8, 9]
target = 65
def BS(arr,low,high,target):
    if low > high:
        return f" element not found"
    mid = (low + high)//2
    if arr[mid] == target:
        return f" element found at index: {mid}."
    elif arr[mid] < target:
        return BS(arr, mid+1,high,target)
    else:
        return BS(arr,low,mid-1,target)

n = len(arr)
BS(arr,0,n-1,target)
```
