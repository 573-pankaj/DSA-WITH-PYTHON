# Lec 46 : Introduction to Binary Search

```python
arr = [1, 2, 3, 4, 5, 6, 7, 8, 9]
target = 68

def BinarySearch(arr, target):
    n = len(arr)
    low = 0
    high = n - 1
    
    # Use a while loop so low and high can shrink inward
    while low <= high:
        mid = (low + high) // 2
        
        # Case 1: Target found at mid index
        if arr[mid] == target:
            return f"Target element found at index {mid}"
            
        # Case 2: If target is SMALLER than mid element, it's in the left half
        # So we move 'high' down to mid - 1
        elif arr[mid] > target:
            high = mid - 1
            
        # Case 3: If target is LARGER than mid element, it's in the right half
        # So we move 'low' up to mid + 1
        else:
            low = mid + 1
            
    return f"Element not found {-1}"

print(BinarySearch(arr, target))
# Output: Target element found at index 5
```

```python
# Recursive Solutions
arr = [1, 2, 3, 4, 5, 6, 7, 8, 9]
target = 65
def BS(arr,low,high,target):
    if low > high:
        return f" element not found"
    mid = (low + high)//2
    if arr[mid] == target:
        return f" element found at index: {mid}."
    elif arr[mid] < target:
        return BS(arr, mid+1,high,target)
    else:
        return BS(arr,low,mid-1,target)

n = len(arr)
BS(arr,0,n-1,target)
```

## Lec 47 : Finding Lower and Upper bound 

```python
arr = [1, 2, 4, 4, 4, 7, 9]
target = 4

def lowerBound(arr, target):
    n = len(arr)
    low = 0
    high = n - 1
    ans = n  # Initialize answer with n (default if target is greater than all elements)
    
    while low <= high:
        mid = (low + high) // 2
        
        # If current element is greater than or equal to target, 
        # it could be our answer, but let's check if there's a smaller valid index on the left.
        if arr[mid] >= target:
            ans = mid      # Store the potential index
            high = mid - 1 # Search in the left half for a tighter match
            
        else:
            # If current element is too small, target must be in the right half
            low = mid + 1
            
    return ans

print(f"Lower Bound index for {target} is: {lowerBound(arr, target)}")
# Output: Lower Bound index for 4 is: 2 (since arr[2] is 4, which is the first occurrence)
```

```python
arr = [1, 2, 4, 4, 4, 7, 9]
target = 4

def upperBound(arr, target):
    n = len(arr)
    low = 0
    high = n - 1
    ans = n  # Default to n if no element is strictly greater than target
    
    while low <= high:
        mid = (low + high) // 2
        
        # If current element is strictly greater than target, 
        # it could be our answer, but let's look for a tighter match on the left.
        if arr[mid] > target:
            ans = mid      # Store potential index
            high = mid - 1 # Search in the left half
            
        else:
            # If current element is less than or equal to target, 
            # the upper bound must be further to the right.
            low = mid + 1
            
    return ans

print(f"Upper Bound index for {target} is: {upperBound(arr, target)}")
# Output: Upper Bound index for 4 is: 5 (since arr[5] is 7, which is the first element strictly > 4)
```

Complexity AnalysisTime Complexity: $O(\log N)$ — Because it cuts the search space in half during every iteration.Space Complexity: $O(1)$ — Uses constant extra space for pointers.
