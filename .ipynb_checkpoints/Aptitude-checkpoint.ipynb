# Lec 6 :  Q.1 Count the number of digits in an integer

<img src="images/for_while_loop2.jpg" alt="Image Description" width="500"/>

```python
num = 123456
num // 10
```

```python
def count_digits(n: int) -> int:
    # 1. Convert negative numbers to positive using abs()
    num = abs(n)
    
    # 2. Edge case: 0 has 1 digit
    if num == 0:
        return 1
        
    count = 0
    
    # 3. Repeatedly divide by 10 until num becomes 0
    while num > 0:
        count += 1
        num = num // 10
        
    return count

# Testing the function
print(count_digits(84438758437563))  # Output: 14
print(count_digits(0))               # Output: 1
print(count_digits(-9875))           # Output: 4
```

# Lec 7 : Check if a number is Palindrome or not

<img src="images/palindrome.jpg" alt="Image Description" width="500"/>

```python
def is_palindrome(n):
    # Negative numbers are not palindromes due to the '-' sign
    if n < 0:
        return False
        
    num = n
    result = 0 
    
    while num > 0:
        digit = num % 10
        result = (result * 10) + digit 
        num = num // 10
        
    # Compare original 'n' with reversed 'result'
    return n == result

# Example Usage:
print(is_palindrome(121))   # True
print(is_palindrome(123))   # False
print(is_palindrome(1221))  # True
print(is_palindrome(-121))  # False
```

```python
def is_palindrome_str(n):
    s = str(n)
    return s == s[::-1]

print(is_palindrome_str(121))  # True
print(is_palindrome(-121))  # False
```

# Lec 8 : Check if the number is an Armstrong number 

```python
153 % 10
```

```python
153 // 10
```

```python
def is_armstrong(n: int) -> bool:
    if n < 0:
        return False
    k = len(str(n))
    num = n    
    total_sum = 0 #= 153
    while num>0:
        
        digit = num % 10 # Extract last digit 
        total_sum += digit**k  # add digit raise to the power k
        num = num // 10       # remove the last digit 
    return total_sum == n
        
```

```python
# Test cases
print(is_armstrong(153))  # True
print(is_armstrong(1634))  # True
print(is_armstrong(371))  # True
print(is_armstrong(123))  # False
```

```python
def is_armstrong(n: int) -> bool:
    # Negative numbers are not Armstrong numbers
    if n < 0:
        return False

    # Convert to string or count via math to find number of digits (k)
    k = len(str(n))

    num = n
    total_sum = 0

    while num > 0:
        digit = num % 10  # Extract last digit
        total_sum += digit**k  # Add digit raised to power k
        num = num // 10  # Remove last digit

    # Return True if sum equals the original number
    return total_sum == n


# Test cases
print(is_armstrong(153))  # True
print(is_armstrong(1634))  # True
print(is_armstrong(371))  # True
print(is_armstrong(123))  # False
```
