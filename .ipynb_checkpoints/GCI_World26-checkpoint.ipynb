```python
class Pokemon:
    def __init__(self, name, type, level):
        self.name = name
        self.type = type
        self.level = level

    def attack(self):
        print(f"{self.name} attacks with {self.type} type!")

    def level_up(self):
        self.level += 1
        print(f"{self.name} leveled up to {self.level}!")
    
# what is  self ? ans : It refers to the instance of the class itself. the current object that is being created or manipulated. It allows you to access the attributes and methods of the class within its own definition.
```

```python
pika = Pokemon("Pikachu", "Electric", 5) # Create an instance of the Pokemon class
pika.attack()  # Output: Pikachu attacks with Electric type!
pika.level_up()  # Output: Pikachu leveled up to 6!
```
