# Lec 46 : Introduction to Binary Search

```python
arr = [1, 2, 3, 4, 5, 6, 7, 8, 9]
target = 68

def BinarySearch(arr, target):
    n = len(arr)
    low = 0
    high = n - 1
    
    # Use a while loop so low and high can shrink inward
    while low <= high:
        mid = (low + high) // 2
        
        # Case 1: Target found at mid index
        if arr[mid] == target:
            return f"Target element found at index {mid}"
            
        # Case 2: If target is SMALLER than mid element, it's in the left half
        # So we move 'high' down to mid - 1
        elif arr[mid] > target:
            high = mid - 1
            
        # Case 3: If target is LARGER than mid element, it's in the right half
        # So we move 'low' up to mid + 1
        else:
            low = mid + 1
            
    return f"Element not found {-1}"

print(BinarySearch(arr, target))
# Output: Target element found at index 5
```

```python
# Recursive Solutions
arr = [1, 2, 3, 4, 5, 6, 7, 8, 9]
target = 65
def BS(arr,low,high,target):
    if low > high:
        return f" element not found"
    mid = (low + high)//2
    if arr[mid] == target:
        return f" element found at index: {mid}."
    elif arr[mid] < target:
        return BS(arr, mid+1,high,target)
    else:
        return BS(arr,low,mid-1,target)

n = len(arr)
BS(arr,0,n-1,target)
```

## Lec 47 : Finding Lower and Upper bound 

```python
arr = [1, 2, 4, 4, 4, 7, 9]
target = 4

def lowerBound(arr, target):
    n = len(arr)
    low = 0
    high = n - 1
    ans = n  # Initialize answer with n (default if target is greater than all elements)
    
    while low <= high:
        mid = (low + high) // 2
        
        # If current element is greater than or equal to target, 
        # it could be our answer, but let's check if there's a smaller valid index on the left.
        if arr[mid] >= target:
            ans = mid      # Store the potential index
            high = mid - 1 # Search in the left half for a tighter match
            
        else:
            # If current element is too small, target must be in the right half
            low = mid + 1
            
    return ans

print(f"Lower Bound index for {target} is: {lowerBound(arr, target)}")
# Output: Lower Bound index for 4 is: 2 (since arr[2] is 4, which is the first occurrence)
```

```python
arr = [1, 2, 4, 4, 4, 7, 9]
target = 4

def upperBound(arr, target):
    n = len(arr)
    low = 0
    high = n - 1
    ans = n  # Default to n if no element is strictly greater than target
    
    while low <= high:
        mid = (low + high) // 2
        
        # If current element is strictly greater than target, 
        # it could be our answer, but let's look for a tighter match on the left.
        if arr[mid] > target:
            ans = mid      # Store potential index
            high = mid - 1 # Search in the left half
            
        else:
            # If current element is less than or equal to target, 
            # the upper bound must be further to the right.
            low = mid + 1
            
    return ans

print(f"Upper Bound index for {target} is: {upperBound(arr, target)}")
# Output: Upper Bound index for 4 is: 5 (since arr[5] is 7, which is the first element strictly > 4)
```

Complexity AnalysisTime Complexity: $O(\log N)$ — Because it cuts the search space in half during every iteration.Space Complexity: $O(1)$ — Uses constant extra space for pointers.

### lec 48 : Leetcode35, Problem Statement
Given a sorted array of distinct integers and a target value, return the index if the target is found. If not, return the index where it would be if it were inserted in order.

```python
from typing import List

class Solution:
    def searchInsert(self, nums: List[int], target: int) -> int:
        n = len(nums)
        low = 0
        high = n - 1
        ans = n  # Default answer if target is greater than all elements
        
        while low <= high:
            mid = (low + high) // 2
            
            # If element is >= target, it's a potential insert position.
            # Record it and look further left for a tighter match.
            if nums[mid] >= target:
                ans = mid
                high = mid - 1
            else:
                # If element is smaller, insert position must be to the right
                low = mid + 1
                
        return ans

arr = [1, 2, 3, 5,6]
target = 7
obj = Solution()
obj.searchInsert(arr,target)
```

# Lec 49 : Find Floor and Ceil Value of a given number

```python
nums = [3,4,4,4,4,8,9,9,10,12,12,14,15]
def getFloorAndCeil(nums,target):
    n= len(nums)
    floor  = -1
    ceil = -1
    low = 0
    high  = n-1
    while low <= high:
        mid = (low + high) //2
        if nums[mid] == target:
            return [nums[mid],nums[mid]]
        elif nums[mid] > target:
            ceil = nums[mid]
            high = mid -1
        else:
            floor = nums[mid]
            low = low + 1
    return [floor, ceil]

target = 6
getFloorAndCeil(nums,target)
```

Complexity AnalysisTime Complexity: $O(\log N)$ — Because the search space is halved in every iteration via Binary Search.   Space Complexity: $O(1)$ — Uses only constant extra space for pointers and variables.   

# Lec 50 : Find First and LAst Occurance is Sorted Array

```python
nums = [1,2,3,3,3,3,3,3,3,6,7,7,78,8]
target = 3
def FirstLast(nums,target):
    n = len(nums)
    first = -1
    last = -1
    for i in range(0,n):
        if nums[i] == target:
            if first == -1:
                first = i
            last = i
    return [first,last]

FirstLast(nums,target)

# Can be solved using lower bound and upper bound method
```

```python
from typing import List

class Solution:
    def lowerBound(self, nums: List[int], target: int) -> int:
        """Finds the first occurrence (modified lower bound)."""
        n = len(nums)
        lb = -1
        low = 0
        high = n - 1
        
        while low <= high:
            mid = (low + high) // 2
            if nums[mid] >= target:
                lb = mid      # Potential first occurrence
                high = mid - 1 # Search left for an earlier occurrence
            else:
                low = mid + 1
                
        return lb

    def upperBound(self, nums: List[int], target: int) -> int:
        """Finds the first element strictly greater than target."""
        n = len(nums)
        ub = -1
        low = 0
        high = n - 1
        
        while low <= high:
            mid = (low + high) // 2
            if nums[mid] > target:
                ub = mid      # Potential upper bound
                high = mid - 1 # Search left for a tighter bound
            else:
                low = mid + 1
                
        return ub

    def searchRange(self, nums: List[int], target: int) -> List[int]:
        """Main method combining lower bound and upper bound logic."""
        lb = self.lowerBound(nums, target)
        
        # If the target doesn't exist in the array
        if lb == -1 or nums[lb] != target:
            return [-1, -1]
            
        ub = self.upperBound(nums, target)
        
        # If ub is -1, it means all elements from lb to the end are equal to target
        last_occurrence = ub - 1 if ub != -1 else len(nums) - 1
        
        return [lb, last_occurrence]

# --- Creating an Object and Executing ---
nums = [5, 7, 7, 8, 8, 10]
target = 8

obj = Solution()
result = obj.searchRange(nums, target)
print(f"First and Last Position: {result}")
# Output: First and Last Position: [3, 4]
```

### Phase 1: Finding the First Occurrence (`lowerBound`)
* **Initial State:** `low = 0`, `high = 5`, `lb = -1`

| Iteration | `low` | `high` | `mid` | `nums[mid]` | Condition (`nums[mid] >= 8`) | Action Taken | Current `lb` |
| :---: | :---: | :---: | :---: | :---: | :---: | :--- | :---: |
| **1** | `0` | `5` | `2` | `7` | `7 >= 8` (False) | `low = mid + 1` $\rightarrow$ `3` | `-1` |
| **2** | `3` | `5` | `4` | `8` | `8 >= 8` (True) | `lb = 4`, `high = mid - 1` $\rightarrow$ `3` | `4` |
| **3** | `3` | `3` | `3` | `8` | `8 >= 8` (True) | `lb = 3`, `high = mid - 1` $\rightarrow$ `2` | `3` |

* **Loop Ends:** `low (3) <= high (2)` is False.
* **Result of `lowerBound`:** Returns **`3`** (`lb = 3`).

---

### Phase 2: Finding the Upper Bound (`upperBound`)
* **Initial State:** `low = 0`, `high = 5`, `ub = -1`

| Iteration | `low` | `high` | `mid` | `nums[mid]` | Condition (`nums[mid] > 8`) | Action Taken | Current `ub` |
| :---: | :---: | :---: | :---: | :---: | :---: | :--- | :---: |
| **1** | `0` | `5` | `2` | `7` | `7 > 8` (False) | `low = mid + 1` $\rightarrow$ `3` | `-1` |
| **2** | `3` | `5` | `4` | `8` | `8 > 8` (False) | `low = mid + 1` $\rightarrow$ `5` | `-1` |
| **3** | `5` | `5` | `5` | `10` | `10 > 8` (True) | `ub = 5`, `high = mid - 1` $\rightarrow$ `4` | `5` |

* **Loop Ends:** `low (5) <= high (4)` is False.
* **Result of `upperBound`:** Returns **`5`** (`ub = 5`).

---

### Phase 3: Final Output Calculation
* **`lb = 3`** (First occurrence of `8`)
* **`ub = 5`** (First element strictly greater than `8`, which is `10` at index `5`)
* **Last occurrence** = `ub - 1` = `5 - 1` = **`4`**
* **Final Returned Range:** `[3, 4]`

# Lec 51 : Count Occurrence in Sorted Array

```python
nums = [1,2,3,3,3,3,3,5,6,8,9,10]
target = 3
def countOccurence(nums,target):
    n = len(nums)
    first = -1
    last = -1
    for i in range(0,n):
        if nums[i] == target:
            if first ==-1:
                first = i
            last = i
    if first == -1:
            return 0
    return last - first + 1

countOccurence(nums,target)
```

### Initial State:

* **`nums`** = `[1, 2, 3, 3, 3, 3, 3, 5, 6, 8, 9, 9, 10]` ($n = 13$)
* **`target`** = `3`
* **`first`** = `-1`, **`last`** = `-1`

### Step-by-Step Iteration

| `i` (Index) | `nums[i]` | Condition (`nums[i] == 3`) | Action Taken | Current `first` | Current `last` |
| :---: | :---: | :---: | :--- | :---: | :---: |
| **0** | `1` | False | None | `-1` | `-1` |
| **1** | `2` | False | None | `-1` | `-1` |
| **2** | `3` | **True** | `first` is -1, so `first = 2`. `last = 2`. | `2` | `2` |
| **3** | `3` | **True** | `first` is not -1. `last = 3`. | `2` | `3` |
| **4** | `3` | **True** | `first` is not -1. `last = 4`. | `2` | `4` |
| **5** | `3` | **True** | `first` is not -1. `last = 5`. | `2` | `5` |
| **6** | `3` | **True** | `first` is not -1. `last = 6`. | `2` | `6` |
| **7** | `5` | False | None | `2` | `6` |
| **8...12**| *varies* | False | None | `2` | `6` |

### Final Calculation

* The loop finishes scanning the array.
* **`first`** is `2` (the index of the first `3`).
* **`last`** is `6` (the index of the last `3`).
* Because `first != -1`, we calculate the total count: `last - first + 1` $\rightarrow$ `6 - 2 + 1` = **`5`**.

**Result:** Returns `5` (Target `3` occurs 5 times in the array).

```python
#Optimal solution

from typing import List

class Solution:
    def lowerBound(self, nums: List[int], target: int) -> int:
        """Finds the first occurrence of the target."""
        n = len(nums)
        lb = -1
        low = 0
        high = n - 1
        
        while low <= high:
            mid = (low + high) // 2
            if nums[mid] >= target:
                lb = mid      
                high = mid - 1 
            else:
                low = mid + 1
                
        return lb

    def upperBound(self, nums: List[int], target: int) -> int:
        """Finds the index of the first element strictly greater than target."""
        n = len(nums)
        ub = n  # Default to n if no element is greater than target
        low = 0
        high = n - 1
        
        while low <= high:
            mid = (low + high) // 2
            if nums[mid] > target:
                ub = mid      
                high = mid - 1 
            else:
                low = mid + 1
                
        return ub

    def countOccurrences(self, nums: List[int], target: int) -> int:
        """Main method to count how many times target appears."""
        lb = self.lowerBound(nums, target)
        
        # If the target doesn't exist in the array
        if lb == -1 or nums[lb] != target:
            return 0
            
        ub = self.upperBound(nums, target)
        
        # The total count is simply Upper Bound index - Lower Bound index
        return ub - lb

# --- Executing ---
# The array MUST be sorted for Binary Search to work!
nums = [1, 2, 5, 7, 7, 8, 8, 10, 12, 12, 12, 12, 12, 12]
nums = sorted(nums)  
# nums becomes: [1, 2, 5, 7, 7, 8, 8, 10, 12, 12, 12, 12, 12, 12] (Length = 14)

target = 12

obj = Solution()
result = obj.countOccurrences(nums, target)
print(f"Target {target} occurs {result} times.")
# Output: Target 12 occurs 6 times.
```

### Initial Setup

* **`nums` (Sorted):** `[1, 2, 5, 7, 7, 8, 8, 10, 12, 12, 12, 12, 12, 12]` ($n = 14$)
* **`target`:** `12`

---

### Phase 1: Finding the First Occurrence (`lowerBound`)

| Iteration | `low` | `high` | `mid` | `nums[mid]` | Condition (`nums[mid] >= 12`) | Action Taken | Current `lb` |
| :---: | :---: | :---: | :---: | :---: | :---: | :--- | :---: |
| **1** | `0` | `13` | `6` | `8` | `8 >= 12` (False) | `low = mid + 1` $\rightarrow$ `7` | `-1` |
| **2** | `7` | `13` | `10` | `12` | `12 >= 12` (True) | `lb = 10`, `high = mid - 1` $\rightarrow$ `9` | `10` |
| **3** | `7` | `9` | `8` | `12` | `12 >= 12` (True) | `lb = 8`, `high = mid - 1` $\rightarrow$ `7` | `8` |
| **4** | `7` | `7` | `7` | `10` | `10 >= 12` (False) | `low = mid + 1` $\rightarrow$ `8` | `8` |
