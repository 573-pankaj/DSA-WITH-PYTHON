# Lec 24 : find lragest value of an Array

```python
def find_largest(arr):
    if not arr:
        return None

    max_val = arr[0]

    for num in arr:
        if num > max_val:
            max_val = num

    return f"Largest element in the array: {max_val}"


arr = [1, 23, 45, 4, 5, 77, 22, 55, 6, 73, 56, 78]
print(find_largest(arr))
# Output: Largest element in the array: 78
```

```python
def find_largest(nums):
    if not nums:
        return None

    largest = nums[0] # 1
    n = len(nums)

    for i in range(0,n):
        largest = max(largest,nums[i])
    return largest


nums = [1, 23, 45, 4, 5, 77, 22, 55, 6, 73, 56, 78]
print(find_largest(nums))
# Output: Largest element in the array: 78

# TC - O(N)
# SC - O(1)
```

```python
def find_largest(nums):
    if not nums:
        return None

    largest = float("-inf")  # set largest as -infinity
    n = len(nums)

    for i in range(0,n):
        largest = max(largest,nums[i])
    return largest


nums = [1, 23, 45, 4, 5, 77, 22, 55, 6, 73, 56, 78]
print(find_largest(nums))
# Output: Largest element in the array: 78

# TC - O(N)
# SC - O(1)
```

### Lec 25 : Find the second largest element of an array

```python
nums = [1, 23, 45, 4, 5, 77, 22, 55, 6, 73, 56, 78]
def find_SecondLargest(nums):
    if not nums:
        return False
    n = len(nums)
    largest = float("-inf")
    s_largest = float("-inf")
    for i in range(0,n):
        if nums[i]>largest:
           
            largest = nums[i]
    for j in range(0,n):
        if nums[j] > s_largest and nums[j] != largest:
            s_largest = nums[j]
    
        
    return largest, s_largest 


# TC - O(N+N) ~ O(N)
# SC - O(1)
```

```python
nums = [1, 23, 45, 4, 5, 77, 22, 55, 6, 73, 56, 78]
print(find_SecondLargest(nums))
```

```python
nums = [1, 23, 45, 4, 5, 77, 22, 55, 6, 73, 56, 78]
def find_SecondLargest(nums):
    if not nums:
        return False
    n = len(nums)
    largest = float("-inf")
    s_largest = float("-inf")
    for i in range(0,n):
        if nums[i]>largest:
           
            s_largest = largest
            largest = nums[i]
        elif nums[i]>s_largest and nums[i] != largest:
            s_largest = nums[i]
        
    return largest, s_largest 

nums = [1, 23, 45, 4, 5, 77, 22, 55, 6, 73, 56, 78]
print(find_SecondLargest(nums))
```

# Lec 26 : Check if an Array is Sorted 

```python
def Sorted(nums):
    for i in range(0,len(nums)-1):
        if nums[i] > nums[i+1]:
            return f"The array is not sorted:  {False}."
    return f" Array is sorted: {True}."
nums = [1,2,34,45,56,6,7432]
Sorted(nums)
            
        
```
