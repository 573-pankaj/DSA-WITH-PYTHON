```python
# # Degree, Path, Cycle, and Connected Graph
#         A
#        / \
#       B   C
#      /     \
#     D-------E

graph = {
    "A": ["B", "C"],
    "B": ["A", "D"],
    "C": ["A", "E"],
    "D": ["B", "E"],
    "E": ["C", "D"]
}
```

```python
# 1. Degree of a Node

# The degree of a node means:

# How many edges are connected to that node?

# Look at A:

#         A
#        / \
#       B   C

# A has two connections:

# A → B
# A → C

# Therefore:

# $$ \text{degree}(A)=2 $$

# In Python:
len(graph["A"]) # degree of note A is 2
```

```python
graph["D"]
```

```python
len(graph["D"])
```

```python
# Key idea
# len(graph[node])

# gives the degree of node for an undirected graph.
```

```python
# 2. Path

# Now suppose we want to go from:

# A → B → D

# We can travel:

# A
# |
# B
# |
# D

# So:

# A → B → D

# is a path from A to D.

# Another path is:

# A → C → E → D

# Therefore, there can be multiple paths between two nodes.

# This idea will become very important in:

# BFS
# DFS
# shortest path algorithms
# network problems
# 3. What Does "Adjacent" Mean?

# Two nodes are adjacent if they have a direct edge between them.

# For example:

# A ----- B

# A and B are adjacent.

# So:

# A and B → adjacent

# But A and D are not directly connected:

# A
# |
# B
# |
# D

# There is no direct:

# A ----- D

# Therefore:

# A and D → not adjacent

# They are connected by a path, but they are not adjacent.
```

```python
# 4. Cycle

# Now look at this part:

# A
# |\
# | \
# B  C
# |  |
# D--E

# We can travel:

# A → B → D → E → C → A

# We started at A and eventually came back to A.

# That is a cycle.

# A cycle is essentially:

# A path that starts and ends at the same vertex.

# For example:

# A → B → D → E → C → A

# is a cycle.
```

```python
# 5. Why Are Cycles Important?

# Because later, when we use DFS, we need to answer questions like:

# "Have I already visited this node?"

# Consider:

# A ----- B
#  \     /
#    \ /
#     C

# Suppose we start at A.

# We can go:

# A → B → C

# But from C, we can see A again.

# If our algorithm doesn't remember that A was already visited, it could keep going:

# A → B → C → A → B → C → A → ...

# forever.

# That's why we use:

# visited = set()

# We'll study this carefully when we reach DFS.
```

```python
# 6. Connected Graph

# Look at our graph:

#         A
#        / \
#       B   C
#      /     \
#     D-------E

# Can we get from A to every other node?

# Yes:

# A → B
# A → C
# A → B → D
# A → C → E

# Therefore, the graph is connected.

# In simple words:


# Every node can be reached from every other node.
```

```python
# 7. Disconnected Graph

# Now consider:

# A ----- B       C ----- D

# There is no connection between the two groups.

# We have two separate components:

# Component 1: A, B

# Component 2: C, D

# So this graph is disconnected.

# This concept will later lead to:

# Connected Components
```

```python
# | Concept                | Simple meaning                            |
# | ---------------------- | ----------------------------------------- |
# | **Node / Vertex**      | A point in the graph                      |
# | **Edge**               | Connection between two nodes              |
# | **Neighbor**           | Node directly connected to another        |
# | **Adjacent**           | Two nodes have a direct edge              |
# | **Degree**             | Number of neighbors                       |
# | **Path**               | Sequence of nodes connected by edges      |
# | **Cycle**              | Path that comes back to its starting node |
# | **Connected graph**    | Every node can reach every other node     |
# | **Disconnected graph** | Some nodes/groups cannot reach each other |
```
