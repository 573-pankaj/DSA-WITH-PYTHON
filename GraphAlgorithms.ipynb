```python
# 1. What is a Graph?

# A graph is simply a collection of:

# Nodes (vertices) → things
# Edges → connections between things

# For example:
    
    
# A ----- B
# |       |
# |       |
# C ----- D
```

```python
# Nodes: A, B, C, D
# Edges: (A, B), (A, C), (B, D), (C, D)
```
