```python
# 2. Path

# Now suppose we want to go from:

# A → B → D

# We can travel:

# A
# |
# B
# |
# D

# So:

# A → B → D

# is a path from A to D.

# Another path is:

# A → C → E → D

# Therefore, there can be multiple paths between two nodes.

# This idea will become very important in:

# BFS
# DFS
# shortest path algorithms
# network problems
# 3. What Does "Adjacent" Mean?

# Two nodes are adjacent if they have a direct edge between them.

# For example:

# A ----- B

# A and B are adjacent.

# So:

# A and B → adjacent

# But A and D are not directly connected:

# A
# |
# B
# |
# D

# There is no direct:

# A ----- D

# Therefore:

# A and D → not adjacent

# They are connected by a path, but they are not adjacent.
```

```python
# 4. Cycle

# Now look at this part:

# A
# |\
# | \
# B  C
# |  |
# D--E

# We can travel:

# A → B → D → E → C → A

# We started at A and eventually came back to A.

# That is a cycle.

# A cycle is essentially:

# A path that starts and ends at the same vertex.

# For example:

# A → B → D → E → C → A

# is a cycle.
```

```python
# 5. Why Are Cycles Important?

# Because later, when we use DFS, we need to answer questions like:

# "Have I already visited this node?"

# Consider:

# A ----- B
#  \     /
#    \ /
#     C

# Suppose we start at A.

# We can go:

# A → B → C

# But from C, we can see A again.

# If our algorithm doesn't remember that A was already visited, it could keep going:

# A → B → C → A → B → C → A → ...

# forever.

# That's why we use:

# visited = set()

# We'll study this carefully when we reach DFS.
```

```python
# 6. Connected Graph

# Look at our graph:

#         A
#        / \
#       B   C
#      /     \
#     D-------E

# Can we get from A to every other node?

# Yes:

# A → B
# A → C
# A → B → D
# A → C → E

# Therefore, the graph is connected.

# In simple words:


# Every node can be reached from every other node.
```

```python
# 1. Define the raw graph data
nodes = ["A", "B", "C", "D"]
edges = [("A", "B"), ("B", "C"), ("C", "D"), ("A", "D")]

# 2. Build the index dictionary (Mapping node name to an integer 0 to N-1)
# This creates: {'A': 0, 'B': 1, 'C': 2, 'D': 3}
index = {node: idx for idx, node in enumerate(nodes)}
print(index)

# 3. Initialize an N x N matrix with zeros
# Note: Using list comprehension ensures we create independent rows in memory
N = len(nodes)
matrix = [[0 for _ in range(N)] for _ in range(N)]
print(f' NxN matrix with zero')
for row in matrix:
    print(row)

# 4. Populate the matrix using your exact loop
for u, v in edges:
    i = index[u]
    j = index[v]

    matrix[i][j] = 1
    matrix[j][i] = 1

# 5. Display the final result
print("Nodes mapped to indices:", index)
print("\nFinal Adjacency Matrix:")
for row in matrix:
    print(row)
```

```python
# 1. Directed Graph (One-Way Connections)
# Edges represent one-way directions (e.g., A -> B)
edges = [("A", "B"), ("B", "C"), ("C", "D"), ("A", "D")]

for u, v in edges:
    i = index[u]
    j = index[v]

    # ONLY create the forward edge
    matrix[i][j] = 1
    # The line `matrix[j][i] = 1` is removed
```

```python
# 2. Weighted Graph (Connections with Values)
# For a weighted graph, your edges need to store the cost, distance, or weight of the connection. You unpack this third value in the loop and assign it to the matrix instead of a hardcoded 1.

# Python
# # Include the weight as a third item in each tuple: (source, destination, weight)
edges = [("A", "B", 5), ("B", "C", 12), ("C", "D", 7), ("A", "D", 3)]

for u, v, weight in edges:
    i = index[u]
    j = index[v]

    # Assign the weight instead of 1
    matrix[i][j] = weight
    matrix[j][i] = weight  # Keep this line if the graph is undirected
```

```python
# BFS algorithm
from collections import deque

def bfs(graph,start):
    visited = {start}
    queue = deque([start])
    
    while queue:              # Keep running as long as there is something waiting in the queue.
        node = queue.popleft()
        print(node)
        
        for neighbor in graph[node]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)
```

```python

# BFS Algo.
class Queue:

    def __init__(self):
        self.data = []
        self.front = 0

    def enqueue(self, value):
        self.data.append(value)

    def dequeue(self):
        if self.front >= len(self.data):
            return None

        value = self.data[self.front]
        self.front += 1

        return value

    def is_empty(self):
        return self.front >= len(self.data)




def bfs(graph, start):

    visited = {start}

    q = Queue()
    q.enqueue(start)

    while not q.is_empty():

        node = q.dequeue()

        print(node)

        for neighbor in graph[node]:

            if neighbor not in visited:

                visited.add(neighbor)
                q.enqueue(neighbor)
```

```python
graph = {
    "A": ["B", "C"],
    "B": ["A", "D"],
    "C": ["A", "E"],
    "D": ["B"],
    "E": ["C"]
}

bfs(graph, "A")
```
