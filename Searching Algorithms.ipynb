### Important Bug Fix Note:

In your original code, you set `high = n`. Because Python lists are **zero-indexed** (meaning a list of length 14 has valid indices from `0` to `13`), setting `high = n` points to an index that does not exist. If you search for a number larger than all elements in the array, `low` will increase until it hits `14`, causing an **`IndexError`**.

To fix this, we always initialize `high = n - 1` so it points to the **last valid element** of the array.

### How Binary Search Works (Step-by-Step Dry Run for `key = 3`)

1. **Initial State:** `low = 0`, `high = 13` (since length is 14)
2. **Iteration 1:**
* `mid = (0 + 13) // 2 = 6`
* `arr[6]` is `57`.
* Is `3 == 57`? No.
* Is `3 < 57`? Yes! So we discard everything from index 6 onwards. We update `high = mid - 1` (`6 - 1 = 5`).


3. **Iteration 2:**
* `low = 0`, `high = 5`
* `mid = (0 + 5) // 2 = 2`
* `arr[2]` is `3`.
* Is `3 == arr[2]`? **Yes!**
* Returns: `"Key element found at index: 2."`



### Complexity Analysis

* **Time Complexity:** $O(\log N)$ — Because the search space is cut in half with every single iteration, it is exceptionally fast even for massive arrays.
* **Space Complexity:** $O(1)$ — It only uses a few pointer variables (`low`, `high`, `mid`), requiring no extra memory.
