# Lec 24 : find lragest value of an Array

```python
def find_largest(arr):
    if not arr:
        return None

    max_val = arr[0]

    for num in arr:
        if num > max_val:
            max_val = num

    return f"Largest element in the array: {max_val}"


arr = [1, 23, 45, 4, 5, 77, 22, 55, 6, 73, 56, 78]
print(find_largest(arr))
# Output: Largest element in the array: 78
```

```python
def find_largest(nums):
    if not nums:
        return None

    largest = nums[0] # 1
    n = len(nums)

    for i in range(0,n):
        largest = max(largest,nums[i])
    return largest


nums = [1, 23, 45, 4, 5, 77, 22, 55, 6, 73, 56, 78]
print(find_largest(nums))
# Output: Largest element in the array: 78

# TC - O(N)
# SC - O(1)
```

### Lec 25 : Find the second largest element of an array

```python
nums = [1, 23, 45, 4, 5, 77, 22, 55, 6, 73, 56, 78]
def find_SecondLargest(nums):
    if not nums:
        return False
    n = len(nums)
    largest = float("-inf")
    s_largest = float("-inf")
    for i in range(0,n):
        if nums[i]>largest:
           
            largest = nums[i]
    for j in range(0,n):
        if nums[j] > s_largest and nums[j] != largest:
            s_largest = nums[j]
    
        
    return largest, s_largest 


# TC - O(N+N) ~ O(N)
# SC - O(1)
```

```python
nums = [1, 23, 45, 4, 5, 77, 22, 55, 6, 73, 56, 78]
print(find_SecondLargest(nums))
```

# Lec 27 Remove Duplicates from a Sroted array 

```python
nums = [0, 0, 1, 1, 1, 2, 2, 3, 3, 4]
nums[0] = 777
print(nums)
```

```python
def removeDuplicates(nums: list[int]) -> int:
    n= len(nums)
    if n==1:
        return 1
    if not nums:
        return 0

    # Pointer 'i' tracks the position of the last placed unique element
    i = 0

    # Pointer 'j' scans the array starting from the second element
    for j in range(1, len(nums)):
        # When a new unique element is found
        if nums[j] != nums[i]:
            i += 1
            nums[i] = nums[j]  # Move unique element into position

    # The number of unique elements is (last_index + 1)
    return i + 1


# --- Test Case ---
nums = [0, 0, 1, 1, 1, 2, 2, 3, 3, 4]
k = removeDuplicates(nums)

print(f"Number of unique elements: {k}")
print(f"Modified array prefix: {nums[:k]}")
print(f"Whole nums : {nums}")
# Output:
# Number of unique elements: 5
# Modified array prefix: [0, 1, 2, 3, 4]

# TC -> O(N)
# SC - > O(1)
```

```python
nums = [1,2,3,4,5,6,7]
def rotate(nums):
    n = len(nums)
    temp = nums[n-1]   # 7
    for i in range(n-2,-1,-1):
        nums[i+1] = nums[i]
    nums[0] = temp
    return nums
rotate(nums)
```

```python
def rotate_right(nums: list[int], k: int) -> None:
    n = len(nums)
    if n == 0:
        return

    # Normalize k in case k > n
    k = k % n

    # Helper function to reverse array in-place between left and right indices
    def reverse(arr, left, right):
        while left < right:
            arr[left], arr[right] = arr[right], arr[left]
            left += 1
            right -= 1

    # Step 1: Reverse the whole array
    reverse(nums, 0, n - 1)

    # Step 2: Reverse the first k elements
    reverse(nums, 0, k - 1)

    # Step 3: Reverse the remaining n - k elements
    reverse(nums, k, n - 1)


# --- Test ---
nums = [1, 2, 3, 4, 5, 6, 7]
rotate_right(nums, 3)
print(nums)
# Output: [5, 6, 7, 1, 2, 3, 4]
```

```python
def rotate_right_slicing(nums: list[int], k: int) -> None:
    n = len(nums)
    if n == 0:
        return
    k %= n

    # Slice last k elements and prepend them to the rest
    nums[:] = nums[-k:] + nums[:-k]

nums = [1, 2, 3, 4, 5, 6, 7]
rotate_right_slicing(nums, 3)
print(nums)
```
