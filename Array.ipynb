```python
def find_largest(arr):
    if not arr:
        return None

    max_val = arr[0]

    for num in arr:
        if num > max_val:
            max_val = num

    return f"Largest element in the array: {max_val}"


arr = [1, 23, 45, 4, 5, 77, 22, 55, 6, 73, 56, 78]
print(find_largest(arr))
# Output: Largest element in the array: 78
```

```python
def find_largest(nums):
    if not nums:
        return None

    largest = nums[0]
    n = len(nums)

    for i in range(0,n):
        largest = max(largest,nums[i])


nums = [1, 23, 45, 4, 5, 77, 22, 55, 6, 73, 56, 78]
print(find_largest(nums))
# Output: Largest element in the array: 78
```
