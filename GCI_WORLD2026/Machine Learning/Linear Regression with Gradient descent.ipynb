# Linear Regression from Scratch — Concepts and Mathematics

This notebook implements **Linear Regression using Gradient Descent**.
The main idea is to learn the model parameters — **weights** \(w\) and **bias** \(b\) — by minimizing the Mean Squared Error (MSE).

---

## 1. Model Initialization — `__init__`

```python
def __init__(self, learning_rate=0.01, epochs=1000):
```

### Hyperparameters

Two important hyperparameters are defined:

* **Learning rate** \(\alpha\): controls the size of each parameter update.
* **Epochs**: number of times the optimization algorithm updates the parameters.

Mathematically,

$$
\alpha = \text{learning rate}
$$

and

$$
T = \text{number of optimization iterations}.
$$

For example:

```python
learning_rate = 0.01
epochs = 1000
```

means that Gradient Descent will perform **1000 parameter updates**, with a step size of \(0.01\).

### Parameter Initialization

```python
self.coef_ = None
self.intercept_ = None
```

These are placeholders for the parameters that will be learned during training.

* `coef_` → feature weights \(w\)
* `intercept_` → bias/intercept \(b\)

For \(n\) features,

$$
w =
\begin{bmatrix}
w_1\\
w_2\\
\vdots\\
w_n
\end{bmatrix}
$$

and the bias is a scalar:

$$
b \in \mathbb{R}.
$$

---

# 2. Training — `fit()`

The `fit()` function learns the optimal values of \(w\) and \(b\) from the training data.

---

## 2.1 Convert Input Data to NumPy Arrays

```python
X_mat = np.asarray(X, dtype=float)
y_vec = np.asarray(y, dtype=float)
```

This converts the input data into NumPy arrays so that mathematical operations such as matrix multiplication can be performed efficiently.

### Shape of the data

Suppose we have:

* \(m\) = number of training samples
* \(n\) = number of features

Then,

$$
X \in \mathbb{R}^{m\times n}
$$

and

$$
y \in \mathbb{R}^{m}.
$$

For example, if we have 100 samples and 3 features:

$$
X \in \mathbb{R}^{100\times3}.
$$

---

## 2.2 Get Number of Samples and Features

```python
m, n = X_mat.shape
```

This gives:

$$
m = \text{number of samples}
$$

$$
n = \text{number of features}.
$$

Therefore,

$$
X =
\begin{bmatrix}
x_{11} & x_{12} & \cdots & x_{1n}\\
x_{21} & x_{22} & \cdots & x_{2n}\\
\vdots & \vdots & \ddots & \vdots\\
x_{m1} & x_{m2} & \cdots & x_{mn}
\end{bmatrix}.
$$

---

# 3. Initialize Model Parameters

```python
self.coef_ = np.zeros(n)
self.intercept_ = 0.0
```

Initially, all weights are set to zero:

$$
w =
\begin{bmatrix}
0\\
0\\
\vdots\\
0
\end{bmatrix}
$$

and

$$
b=0.
$$

The model initially predicts:

$$
\hat{y}=0.
$$

These are only starting values. Gradient Descent will update them during training.

---

# 4. Linear Regression Model

For one sample, the prediction is

$$
\hat{y}
=
w_1x_1+w_2x_2+\cdots+w_nx_n+b.
$$

In vector notation,

$$
\boxed{\hat{y}=x^Tw+b}
$$

For all \(m\) samples simultaneously,

$$
\boxed{\hat{y}=Xw+b}
$$

where \(b\) is added to every sample.

In Python:

```python
y_pred = X_mat @ self.coef_ + self.intercept_
```

Here, `@` represents **matrix multiplication**.

---

# 5. Gradient Descent Training Loop

```python
for _ in range(self.epochs):
```

This loop repeatedly updates \(w\) and \(b\).

If

```python
epochs = 1000
```

then the model performs 1000 iterations of Gradient Descent.

The goal is to find parameters \(w\) and \(b\) that minimize the loss function.

---

# 6. Prediction

```python
y_pred = X_mat @ self.coef_ + self.intercept_
```

The model calculates predictions using

$$
\boxed{\hat{y}=Xw+b}.
$$

For example, for two features,

$$
\hat{y}
=
w_1x_1+w_2x_2+b.
$$

For \(m\) samples, all predictions are calculated simultaneously using matrix multiplication.

---

# 7. Calculate the Error

```python
error = y_pred - y_vec
```

The prediction error for sample \(i\) is

$$
e_i=\hat{y}_i-y_i.
$$

Therefore,

$$
\boxed{e=\hat{y}-y}.
$$

The error tells us how far the prediction is from the actual target.

### Example

If

$$
y=10
$$

and

$$
\hat{y}=8,
$$

then

$$
e=8-10=-2.
$$

If

$$
y=10
$$

and

$$
\hat{y}=12,
$$

then

$$
e=12-10=2.
$$

---

# 8. Mean Squared Error (MSE)

The loss function used by Linear Regression is usually the **Mean Squared Error**:

$$
\boxed{
J(w,b)
=
\frac{1}{m}
\sum_{i=1}^{m}
(\hat{y}_i-y_i)^2
}
$$

Using

$$
\hat{y}_i=w^Tx_i+b,
$$

we can write

$$
J(w,b)
=
\frac{1}{m}
\sum_{i=1}^{m}
(w^Tx_i+b-y_i)^2.
$$

The goal of Gradient Descent is:

$$
\boxed{
\min_{w,b} J(w,b)
}
$$

---

# 9. Gradient with Respect to the Weights

```python
dw = (1/m) * (X_mat.T @ error)
```

The gradient of the MSE with respect to the weights is

$$
\boxed{
\frac{\partial J}{\partial w}
=
\frac{1}{m}X^T(\hat{y}-y)
}
$$

Since

$$
e=\hat{y}-y,
$$

we can write

$$
\boxed{
dw=\frac{1}{m}X^Te
}
$$

### Why \(X^T\)?

The matrix \(X\) has shape

$$
m\times n.
$$

Therefore,

$$
X^T
$$

has shape

$$
n\times m.
$$

Since the error vector has shape

$$
m\times1,
$$

we get

$$
(n\times m)(m\times1)
=
n\times1.
$$

This gives one gradient value for every feature weight.

---

# 10. Gradient with Respect to the Bias

```python
db = (1/m) * np.sum(error)
```

The derivative of the MSE with respect to the bias is

$$
\boxed{
\frac{\partial J}{\partial b}
=
\frac{1}{m}
\sum_{i=1}^{m}
(\hat{y}_i-y_i)
}
$$

Therefore,

$$
\boxed{
db=\frac{1}{m}\sum e
}
$$

The bias has only one parameter, so its gradient is a scalar.

---

# 11. Gradient Descent Parameter Update

After calculating the gradients, we update the parameters:

```python
self.coef_ -= self.lr * dw
self.intercept_ -= self.lr * db
```

Mathematically,

$$
\boxed{
w \leftarrow w-\alpha\frac{\partial J}{\partial w}
}
$$

and

$$
\boxed{
b \leftarrow b-\alpha\frac{\partial J}{\partial b}
}
$$

where

$$
\alpha=\text{learning rate}.
$$

### Why subtract the gradient?

The gradient points in the direction of **steepest increase** of the loss.

Therefore, moving in the opposite direction,

$$
-\nabla J,
$$

moves us toward a lower loss.

Hence,

$$
\boxed{\text{New parameter}
=
\text{Old parameter}
-
\text{Learning rate}\times\text{Gradient}}
$$

---

# 12. Complete Gradient Descent Process

Each iteration follows these steps:

$$
\boxed{
X,w,b
\rightarrow
\hat{y}
\rightarrow
e
\rightarrow
dw,db
\rightarrow
w,b\text{ update}
}
$$

More explicitly:

### Step 1 — Predict

$$
\hat{y}=Xw+b
$$

### Step 2 — Calculate error

$$
e=\hat{y}-y
$$

### Step 3 — Calculate weight gradient

$$
dw=\frac{1}{m}X^Te
$$

### Step 4 — Calculate bias gradient

$$
db=\frac{1}{m}\sum e
$$

### Step 5 — Update parameters

$$
w\leftarrow w-\alpha dw
$$

$$
b\leftarrow b-\alpha db
$$

These steps are repeated for the specified number of epochs.

---

# 13. Prediction — `predict()`

```python
return X_mat @ self.coef_ + self.intercept_
```

Once training is complete, the learned \(w\) and \(b\) are used to make predictions on new data.

The prediction equation is

$$
\boxed{
\hat{y}=Xw+b
}
$$

For a single observation,

$$
\boxed{
\hat{y}=w_1x_1+w_2x_2+\cdots+w_nx_n+b
}
$$

The important point is that **no parameter update happens during prediction**.

The model simply uses the learned parameters.

---

# 14. Model Evaluation — \(R^2\) Score

The model can be evaluated using the **Coefficient of Determination**, denoted by

$$
R^2.
$$

The implementation first calculates the **Residual Sum of Squares (RSS)**:

```python
ss_res = np.sum((y_true - y_pred)**2)
```

Mathematically,

$$
\boxed{
SS_{\text{res}}
=
\sum_{i=1}^{m}(y_i-\hat{y}_i)^2
}
$$

This measures the amount of error that remains after using the model.

---

## 14.1 Total Sum of Squares

```python
ss_tot = np.sum((y_true - np.mean(y_true))**2)
```

Mathematically,

$$
\boxed{
SS_{\text{tot}}
=
\sum_{i=1}^{m}(y_i-\bar{y})^2
}
$$

where

$$
\bar{y}
=
\frac{1}{m}\sum_{i=1}^{m}y_i
$$

is the mean of the actual target values.

\(SS_{\text{tot}}\) represents the total variation in the target variable.

---

## 14.2 Formula for \(R^2\)

Finally,

```python
return 1 - (ss_res / ss_tot)
```

implements

$$
\boxed{
R^2
=
1-\frac{SS_{\text{res}}}{SS_{\text{tot}}}
}
$$

Therefore,

$$
\boxed{
R^2
=
1-
\frac{
\sum_{i=1}^{m}(y_i-\hat{y}_i)^2
}{
\sum_{i=1}^{m}(y_i-\bar{y})^2
}
}
$$

### Interpretation

* \(R^2=1\) → predictions perfectly match the actual values.
* \(R^2=0\) → the model performs like simply predicting the mean \(\bar y\).
* \(R^2<0\) → the model performs worse than the mean-baseline predictor.

---

# 15. Data Preprocessing

Before training the model, the dataset is usually divided into training and test sets.

```python
X_train, X_test, y_train, y_test = train_test_split(...)
```

The purpose is to evaluate the model on data that was **not used for training**.

For example, with a 50/50 split:

$$
50\% \rightarrow \text{Training Data}
$$

$$
50\% \rightarrow \text{Test Data}
$$

The model learns \(w\) and \(b\) only from the training data.

The test data is then used to evaluate how well the learned model generalizes to unseen data.

---

# 16. Feature Scaling — Standardization

Gradient Descent works better when numerical features have comparable scales.

A common approach is **Standardization**.

The formula is

$$
\boxed{
z=\frac{x-\mu}{\sigma}
}
$$

where:

* \(x\) = original feature value
* \(\mu\) = mean of the feature
* \(\sigma\) = standard deviation
* \(z\) = standardized value

After standardization, the feature approximately has

$$
\text{mean}=0
$$

and

$$
\text{standard deviation}=1.
$$

---

## 16.1 Fit the Scaler on Training Data

```python
X_train_scaled = scaler.fit_transform(X_train)
```

`fit` calculates the training-data statistics:

$$
\mu_{\text{train}},\quad \sigma_{\text{train}}.
$$

Then `transform` uses these statistics to standardize the training data.

---

## 16.2 Transform the Test Data

```python
X_test_scaled = scaler.transform(X_test)
```

The test data must be transformed using the **same** mean and standard deviation obtained from the training data:

$$
\boxed{
z_{\text{test}}
=
\frac{x_{\text{test}}-\mu_{\text{train}}}
{\sigma_{\text{train}}}
}
$$

We do **not** calculate a new mean and standard deviation from the test set.

This prevents **data leakage**.

---

# 17. Why Feature Scaling Helps Gradient Descent

Suppose we have two features:

$$
x_1 \in [0,10]
$$

and

$$
x_2 \in [0,1,000,000].
$$

The second feature has a much larger numerical scale.

Without scaling, the optimization landscape can become poorly conditioned, causing Gradient Descent to take inefficient steps.

A simplified intuition is:

$$
\text{Different feature scales}
\quad\Rightarrow\quad
\text{uneven gradient updates}
$$

whereas

$$
\text{Similar feature scales}
\quad\Rightarrow\quad
\text{more stable and efficient optimization}.
$$

Therefore, standardization is especially useful when training Linear Regression using Gradient Descent.

# 20. One-Line Intuition

> **Linear Regression learns the best weights \(w\) and bias \(b\) by repeatedly predicting the output, measuring the error, calculating the gradients, and moving the parameters in the direction that reduces the Mean Squared Error.**


````python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler


class LinearRegressionGD:

  def __init__(self, learning_rate=0.01, epochs=1000):
    self.lr = learning_rate
    self.epochs = epochs
    self.coef_ = None
    self.intercept_ = None

  def fit(self, X, y):
    X_mat = np.asarray(X, dtype=float)
    y_vec = np.asarray(y, dtype=float)
    m, n = X_mat.shape

    # Initialize weight and bias to zero
    self.coef_ = np.zeros(n)
    self.intercept_ = 0.0

    # Gradient descent Optimization loop
    for _ in range(self.epochs):
      # Linear model prediction: Y_pred = X*w + b
      y_pred = X_mat @ self.coef_ + self.intercept_

      # Calculate Gradients
      error = y_pred - y_vec
      dw = (1 / m) * (X_mat.T @ error)
      db = (1 / m) * np.sum(error)

      # Update parameters
      self.coef_ -= self.lr * dw
      self.intercept_ -= self.lr * db

    return self

  def predict(self, X):
    X_mat = np.asarray(X, dtype=float)
    return X_mat @ self.coef_ + self.intercept_

  def score(self, X, y):
    y_true = np.asarray(y, dtype=float)
    y_pred = self.predict(X)

    ss_res = np.sum((y_true - y_pred) ** 2)
    ss_tot = np.sum((y_true - np.mean(y_true)) ** 2)

    # ADDED: Explicitly return the calculated R^2 score
    return 1 - (ss_res / ss_tot)


# --- Toy Dataset Creation ---
data = {
    'Size': [1000, 1200, 1500, 1800, 2000, 2200],
    'Bedrooms': [2, 2, 3, 3, 4, 4],
    'Price': [300, 340, 400, 470, 500, 560],
}
X = pd.DataFrame(data)[['Size', 'Bedrooms']]
y = pd.Series(data['Price'])

# Train/Test split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.5, random_state=0
)

# Feature Scaling (Crucial for Gradient Descent)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Train Gradient Descent model
model = LinearRegressionGD(learning_rate=0.05, epochs=2000)
model.fit(X_train_scaled, y_train)

# Evaluate results
print('R^2 (train): {:.3f}'.format(model.score(X_train_scaled, y_train)))
print('R^2 (test): {:.3f}'.format(model.score(X_test_scaled, y_test)))
print(
    '\nCoefficients:\n{}'.format(
        pd.Series(model.coef_, index=X.columns)
    )
)
print('Intercept: {:.3f}'.format(model.intercept_))
````
