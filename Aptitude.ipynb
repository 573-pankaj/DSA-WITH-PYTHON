```python
def count_digits(n: int) -> int:
    # 1. Convert negative numbers to positive using abs()
    num = abs(n)
    
    # 2. Edge case: 0 has 1 digit
    if num == 0:
        return 1
        
    count = 0
    
    # 3. Repeatedly divide by 10 until num becomes 0
    while num > 0:
        count += 1
        num = num // 10
        
    return count

# Testing the function
print(count_digits(84438758437563))  # Output: 14
print(count_digits(0))               # Output: 1
print(count_digits(-9875))           # Output: 4
```

# Lec 7 : Check if a number is Palindrome or not

<img src="images/palindrome.jpg" alt="Image Description" width="500"/>

```python
def is_palindrome(n):
    # Negative numbers are not palindromes due to the '-' sign
    if n < 0:
        return False
        
    num = n
    result = 0 
    
    while num > 0:
        digit = num % 10
        result = (result * 10) + digit 
        num = num // 10
        
    # Compare original 'n' with reversed 'result'
    return n == result

# Example Usage:
print(is_palindrome(121))   # True
print(is_palindrome(123))   # False
print(is_palindrome(1221))  # True
print(is_palindrome(-121))  # False
```

# Lec 8 : Check if the number is an Armstrong number 

An Armstrong number (also known as a narcissistic number) is a number that is equal to the sum of its own digits each raised to the power of the total number of digits. 
For example:153 (3 digits): $1^3 + 5^3 + 3^3 = 1 + 125 + 27 = 153$ $\rightarrow$ Armstrong Number1634 (4 digits): $1^4 + 6^4 + 3^4 + 4^4 = 1 + 1296 + 81 + 256 = 1634$ $\rightarrow$  Armstrong Number123 (3 digits): $1^3 + 2^3 + 3^3 = 1 + 8 + 27 = 36 \neq 123$ $\rightarrow$ Not an Armstrong NumberAlgorithm LogicFind the total number of digits ($k$) in the input integer $n$.Extract each digit one by one using modulo arithmetic (digit = num % 10).Raise each extracted digit to the power of $k$ and add it to a running sum.Remove the processed digit using integer division (num = num // 10).Compare the calculated sum with the original number $n$

```python
def is_armstrong(n: int) -> bool:
    if n < 0:
        return False
    k = len(str(n))
    num = n    
    total_sum = 0 #= 153
    while num>0:
        
        digit = num % 10 # Extract last digit 
        total_sum += digit**k  # add digit raise to the power k
        num = num // 10       # remove the last digit 
    return total_sum == n
        
```

```python
def is_armstrong(n: int) -> bool:
    # Negative numbers are not Armstrong numbers
    if n < 0:
        return False

    # Convert to string or count via math to find number of digits (k)
    k = len(str(n))

    num = n
    total_sum = 0

    while num > 0:
        digit = num % 10  # Extract last digit
        total_sum += digit**k  # Add digit raised to power k
        num = num // 10  # Remove last digit

    # Return True if sum equals the original number
    return total_sum == n


# Test cases
print(is_armstrong(153))  # True
print(is_armstrong(1634))  # True
print(is_armstrong(371))  # True
print(is_armstrong(123))  # False
```

```python
armstrong_numbers = []

# Range of all 3-digit numbers
for num in range(100, 1000):
    temp = num    # 371

    # Pure math digit extraction
    ones = temp % 10    # 1
    temp //= 10         #37

    tens = temp % 10   # 7
    hundreds = temp // 10 # 3

    # For 3-digit numbers, each digit is raised to power 3
    if (hundreds**3 + tens**3 + ones**3) == num:
        armstrong_numbers.append(num)

print("3-digit Armstrong numbers:", armstrong_numbers)
```

```python
def all_factors(n):
    result = []
    i = 1
    
    # Iterate only up to the square root of n
    while i * i <= n:
        if n % i == 0:
            result.append(i)
            # If the paired factor (n // i) is different, add it too
            if i != n // i:
                result.append(n // i)
        i += 1
        
    return sorted(result)

# Test
print(all_factors(10))  # Output: [1, 2, 5, 10]

# time complexity = O(sqrt(n)) + O(nlog(n))
# space complexity = O(k)
```

# Lec 10 : Store Frequency Dictonary

A frequency dictionary (also known as a frequency map or hash map) tracks how many times each unique item appears within a dataset (string, list, tuple, etc.).

In Python, keys store the unique elements, and values store their respective occurrence counts.

```python
data = ["apple", "banana", "apple", "orange", "banana", "apple"]

freq = {}
for item in data:
    freq[item] = freq.get(item, 0) + 1

print(freq)
# Output: {'apple': 3, 'banana': 2, 'orange': 1}
```

```python
from collections import defaultdict

data = "abracadabra"

freq = defaultdict(int)
for char in data:
    freq[char] += 1

print(dict(freq))
# Output: {'a': 5, 'b': 2, 'r': 2, 'c': 1, 'd': 1}
```

```python
from collections import Counter

data = [1, 2, 2, 3, 3, 3, 4, 4, 4, 4]

# Built directly in O(N) time
freq = Counter(data)

print(freq)
# Output: Counter({4: 4, 3: 3, 2: 2, 1: 1})
```

```python
freq = {"apple": 3, "banana": 2, "orange": 1}

most_frequent = None
max_count = 0

for item, count in freq.items():
    if count > max_count:
        max_count = count
        most_frequent = item

print(f"Most frequent: '{most_frequent}' with count {max_count}")
# Output: Most frequent: 'apple' with count 3
```
