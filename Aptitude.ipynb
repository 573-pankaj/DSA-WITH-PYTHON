```python
def count_digits(n: int) -> int:
    # 1. Convert negative numbers to positive using abs()
    num = abs(n)
    
    # 2. Edge case: 0 has 1 digit
    if num == 0:
        return 1
        
    count = 0
    
    # 3. Repeatedly divide by 10 until num becomes 0
    while num > 0:
        count += 1
        num = num // 10
        
    return count

# Testing the function
print(count_digits(84438758437563))  # Output: 14
print(count_digits(0))               # Output: 1
print(count_digits(-9875))           # Output: 4
```
