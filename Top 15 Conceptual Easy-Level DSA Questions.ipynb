### 1 Move Zeroes(LeetCode283)

```python
arr = [1,0,3,0,4,5]
def moveZeros(arr):
    write = 0
    n = len(arr)
    for i in range(0,n):
        if arr[i] !=0:
            arr[write] ,arr[i] = arr[i],arr[write] 
            write +=1
    return arr

moveZeros(arr)
            
```
