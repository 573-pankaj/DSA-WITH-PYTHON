```python
def reverse_string(s):
    result = ""
    for i in range(len(s) - 1, -1, -1):
        result += s[i]
    return result


print(reverse_string("abcd"))    # dcba
print(reverse_string("hello"))   # olleh
print(reverse_string("Python"))  # nohtyP
print(reverse_string(""))        # (empty string — no error)
print(reverse_string("a"))       # a
```

# Ques 2 : Palindrome 

```python
def is_palindrome(s):
    left = 0
    right = len(s) - 1

    while left < right:
        if s[left] != s[right]:
            return False

        left += 1
        right -= 1

    return True
```

```python
is_palindrome("pankaj")
```

```python
print(is_palindrome(""))
```

```python
print(is_palindrome("a"))
```

```python
print(is_palindrome("madam"))
```

```python
print(is_palindrome("abba"))
```

```python
print(is_palindrome("nurses run"))
```

"How would you check whether a string is a palindrome?"

A strong beginner-friendly answer would be:

"A palindrome reads the same forward and backward. One simple approach is to reverse the string and compare it with the original, which takes O(n) time and O(n) extra space. A more space-efficient approach is to use two pointers: one at the beginning and one at the end. I compare those characters and move both pointers toward the center. If I ever find a mismatch, I return false; otherwise, after reaching the center, I return true. This takes O(n) time and O(1) extra space."

That is the kind of explanation you should aim to give naturally.

# Question 3 : FizzBuzz

```python
def fizz_buzz(n):
    for i in range(1, n + 1):

        if i % 3 == 0 and i % 5 == 0:
            print("FizzBuzz")

        elif i % 3 == 0:
            print("Fizz")

        elif i % 5 == 0:
            print("Buzz")

        else:
            print(i)
```

```python
fizz_buzz(15)
```

Time Complexity = O(n)

O(n) time \
O(1) extra space

```python
fizz_buzz(1)
```

```python
fizz_buzz(3)
```

```python
fizz_buzz(0)
```

🧠 Your Interview Explanation

Practice saying this:

"I need to process every number from 1 to n, so I'll use a loop. For each number, I first check whether it is divisible by both 3 and 5, because that's the overlapping case and must be handled first. Then I check divisibility by 3, then 5, otherwise I print the number. The modulo operator lets me test divisibility by checking whether the remainder is zero. The time complexity is O(n) and the extra space is O(1)."

That is a strong answer for a beginner-level interview.

The Problem Statement — In Layman's Terms

listen

```python
def are_anagrams(s1, s2):
    return sorted(s1) == sorted(s2)
```

```python
def are_anagrams1(s1, s2):
    count1 = {}
    count2 = {}

    for ch in s1:
        if ch in count1:
            count1[ch] += 1
        else:
            count1[ch] = 1

    for ch in s2:
        if ch in count2:
            count2[ch] += 1
        else:
            count2[ch] = 1

    return count1 == count2
```

```python
def are_anagrams2(s1, s2):
    if len(s1) != len(s2):
        return print(" length miss match")

    counts = {}

    for ch in s1:
        if ch in counts:
            counts[ch] += 1
        else:
            counts[ch] = 1

    for ch in s2:
        if ch not in counts:
            return print("charachter is not present that means its not anagrams")

        counts[ch] -= 1

        if counts[ch] < 0:
            return print("This catches the frequency mismatch")

    return True
```
